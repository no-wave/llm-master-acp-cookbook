# 2장. CrewAI: Tutorial

## CrewAI란?

CrewAI는 자율적인 AI 에이전트를 조정하여 복잡한 작업을 처리하도록 돕는 프레임워크이다. 각 \*\*에이전트(Agent)\*\*는 특정 역할과 목표를 가지며, 주어진 \*\*작업(Task)\*\*을 수행한다. 이러한 에이전트와 작업의 묶음을 \*\*크루(Crew)\*\*라고 부르며, 크루는 설정된 프로세스에 따라 작업을 체계적으로 실행한다.

## **튜토리얼 1: 기본 에이전트 크루 만들기**

### 1단계: 라이브러리 임포트 및 에이전트 정의

필요한 모듈을 불러오고, 각자의 역할과 목표를 가진 두 에이전트를 만든다.

```python
from crewai import Agent, Task, Crew, Process

# 시장 조사 전문가 에이전트 정의
researcher = Agent(
  role='수석 시장 분석가',
  goal='최신 AI 트렌드에 대한 상세한 분석 보고서 작성',
  backstory="""당신은 기술, 특히 AI 분야의 시장 동향을 파악하는 데 탁월한 능력을 가진 분석가입니다.
  복잡한 데이터를 이해하기 쉽게 설명하는 데 능숙합니다.""",
  verbose=True,
  allow_delegation=False
)

# 기술 전문 작가 에이전트 정의
writer = Agent(
  role='기술 콘텐츠 작가',
  goal='AI 트렌드에 대한 흥미로운 블로그 게시물 작성',
  backstory="""당신은 복잡한 기술 주제를 일반 대중이 쉽게 이해할 수 있도록 설명하는 데 재능이 있는 작가입니다.
  매력적이고 흡입력 있는 글쓰기 스타일을 가지고 있습니다.""",
  verbose=True,
  allow_delegation=True
)
```

### 2단계: 작업(Tasks) 정의

에이전트들이 수행할 구체적인 임무를 정의한다.

```python
# 조사 작업 정의
research_task = Task(
  description="""2024년 최신 AI 기술 트렌드 3가지를 식별하고, 각 트렌드에 대한 핵심 정보를 요약합니다.
  주요 플레이어, 기술적 혁신, 그리고 시장 전망을 포함해야 합니다.""",
  expected_output="각 AI 트렌드에 대한 상세 정보가 담긴 글머리 기호 목록 형식의 보고서.",
  agent=researcher
)

# 작성 작업 정의
write_task = Task(
  description="""조사된 AI 트렌드 분석 보고서를 바탕으로, 일반 독자들을 위한 흥미로운 블로그 게시물을 작성합니다.
  전문 용어 사용을 최소화하고, 실생활 예시를 들어 설명해주세요.""",
  expected_output="AI 최신 트렌드에 대한 내용이 담긴 500단어 내외의 블로그 게시물.",
  agent=writer
)
```

### 3단계: 크루(Crew) 구성 및 실행

에이전트와 작업을 모아 크루를 구성하고, 작업을 시작한다.

```python
# 크루 구성
basic_crew = Crew(
  agents=[researcher, writer],
  tasks=[research_task, write_task],
  process=Process.sequential,  # 작업은 순서대로 진행됩니다 (조사 -> 작성)
  verbose=True
)

# 크루 작업 시작!
basic_result = basic_crew.kickoff()

print("\n######################")
print("## 기본 크루 작업 결과 ##")
print(basic_result)
```

## 튜토리얼 2: RAG 에이전트 크루 만들기

이제 특정 웹 페이지의 정보를 학습하여 질문에 답변하는 RAG(Retrieval-Augmented Generation) 에이전트를 만들어 보겠습니다.

## 1단계: RAG 도구(Tool) 및 에이전트 준비

`RagTool`을 사용하여 CrewAI 공식 문서 페이지를 지식 베이스로 만들고, 이 도구를 사용할 전문가 에이전트를 정의합니다.

```python
from crewai_tools import RagTool

# RAG 도구 초기화 및 데이터 소스 추가
docs_rag_tool = RagTool(
    config={
        "llm": {
            "provider": "openai",
            "config": {
                "model": "gpt4o-mini",
            },
        },
        "embedder": {
            "provider": "openai",
            "config": {
                "model": "text-embedding-3-small",
            },
        },
    }
)
# RAG 소스 추가하기는 add 메서드 사용
docs_rag_tool.add("https://docs.crewai.com/en/core-concepts/agents/")


# CrewAI 문서 전문가 에이전트 정의
docs_qa_agent = Agent(
    role="CrewAI 문서 전문가",
    goal="CrewAI 프레임워크에 대한 사용자의 질문에 정확하게 답변합니다.",
    backstory="""당신은 CrewAI 프레임워크의 공식 문서를 완벽하게 숙지하고 있는 전문가입니다.
    사용자가 CrewAI의 개념이나 사용법에 대해 질문하면, 문서를 기반으로 명확하고 정확한 답변을 제공합니다.""",
    tools=[docs_rag_tool], # RAG 도구를 에이전트에게 부여합니다.
    verbose=True
)
```

### 2단계: RAG 작업 정의 및 크루 실행

RAG 에이전트가 수행할 작업을 정의하고, 새로운 크루를 구성하여 실행한다.

```python
# RAG 작업 정의
rag_task = Task(
    description="CrewAI에서 Agent의 'backstory'는 어떤 역할을 하는지 설명해줘.",
    expected_output="'backstory'의 역할과 중요성에 대한 상세하고 이해하기 쉬운 설명.",
    agent=docs_qa_agent
)

# RAG 크루 구성 및 실행
rag_crew = Crew(
    agents=[docs_qa_agent],
    tasks=[rag_task],
    verbose=True
)

# 크루 작업 시작
rag_result = rag_crew.kickoff()

print("\n######################")
print("## RAG 크루 작업 결과 ##")
print(rag_result)
```
